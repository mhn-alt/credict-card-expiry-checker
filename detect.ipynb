```python
import cv2
import pytesseract
import pandas as pd 
import re
from datetime import datetime
```

```python
img=cv2.imread('cr2.jpg')
img
```

```python
cv2.imshow('cr2.jpg',img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
pytesseract.pytesseract.tesseract_cmd=r"C:\Program Files\Tesseract-OCR\tesseract.exe"
```

```python
text=pytesseract.image_to_string(img)
print(text)
```

```python
data=pytesseract.image_to_data(img,output_type=pytesseract.Output.DICT)
print(data)
```

```python
for i in range(len(data['text'])):
    result=re.search(r'(?<!\d)(0[1-9]|1[0-2])\s*[/\-]\s*(\d{2}|\d{4})(?!\d)',data['text'][i])
    if result:
        print(data['text'][i])
        x=data['left'][i]
        y=data['top'][i]
        w=data['width'][i]
        h=data['height'][i]
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)
        expiry_date = data['text'][i].strip()
        expiry = datetime.strptime(expiry_date, "%m/%y")
        today = datetime.today()
        if expiry.year < today.year or (expiry.year == today.year and expiry.month < today.month):
            print("Card Expired")
        else:
            print("Card is Valid")
cv2.imshow('img',img)
cv2.waitKey()
cv2.destroyAllWindows()        
```
